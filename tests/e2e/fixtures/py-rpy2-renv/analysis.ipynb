An rpy2 notebook whose R packages an renv.lock pins. base-py has no renv, so the lockfile is restored in base-r; the pinned version (not CRAN's current one) must load in base-py's R. fastmatch has compiled code, so a library built for the other image's R would fail here.

```python
%load_ext rpy2.ipython
import os

os.makedirs('results', exist_ok=True)
```

```python
%%R
library(fastmatch)
stopifnot(fmatch(3L, 1:5) == 3L)
writeLines(as.character(packageVersion('fastmatch')), 'results/version.txt')
```
